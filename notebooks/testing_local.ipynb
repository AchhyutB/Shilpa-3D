```python
import json
import os

sdir = '/content/drive/MyDrive/shilpa3D/sessions/test004'

# Check current status
if os.path.exists(f'{sdir}/status.json'):
    status = json.load(open(f'{sdir}/status.json'))
    print(f"Current stage: {status['stage']}")
    print(f"Progress: {status['percent']}%")
else:
    print("No status.json yet")

# Check what directories exist
if os.path.exists(sdir):
    print(f"\nContents of {sdir}:")
    for item in os.listdir(sdir):
        path = os.path.join(sdir, item)
        if os.path.isdir(path):
            files = len(os.listdir(path))
            print(f"  {item}/ ({files} items)")
        else:
            print(f"  {item}")

# Check if there's a results folder with manifest
if os.path.exists(f'{sdir}/results/manifest.json'):
    print("\n✓ PIPELINE COMPLETE!")
    print(json.load(open(f'{sdir}/results/manifest.json')))
```

```python
import subprocess
import os

os.chdir('/content')
subprocess.run(['rm', '-rf', 'gaussian-splatting'], capture_output=True)

# Clone
subprocess.run(['git', 'clone', '--recursive',
                'https://github.com/graphdeco-inria/gaussian-splatting'], check=True)

os.chdir('/content/gaussian-splatting')

# Install submodules and main package
# Upgrade pip and setuptools for better compatibility with complex builds
print("Upgrading pip and setuptools...")
subprocess.run(['pip', 'install', '--upgrade', 'pip', 'setuptools'], check=True, capture_output=True, text=True)
print("Installing cmake and ninja...")
subprocess.run(['pip', 'install', 'cmake', 'ninja'], check=True, capture_output=True, text=True)

print("Installation of core Gaussian Splatting dependencies is handled in separate cells.")
# The main gaussian-splatting directory is not a pip-installable package.
# Its Python dependencies are installed from their submodules (e.g., diff-gaussian-rasterization, simple-knn).
# The top-level scripts are run directly from the cloned repository.

print("✓ All installations complete")
```
